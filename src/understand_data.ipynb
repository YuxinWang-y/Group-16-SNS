```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
file_rotterdamPortData = r'..\data\rotterdam_dataset_16.xlsx'
wd = pd.read_excel(file_rotterdamPortData)
wd.info()
wd.head()
```

```python
# check if Categories of Vessels are balanced
counts = wd['Vessel_Type'].value_counts()
shares = (counts / counts.sum() * 100).round(2)

composition = pd.DataFrame({'Count': counts, 'Share(%)': shares})
composition.loc['Total'] = [composition['Count'].sum(), composition['Share(%)'].sum()]
print(composition)
# From the result, we can forsee that N(t) decomposition will be container-dominated
```

```python
wd.loc[2]
wd.describe()
print(wd[['Cargo_tons', 'Berth_Time_hr']].describe())
print("Date range:", wd['Arrival_Date'].min(), "->", wd['Arrival_Date'].max())
print("Span:", wd['Arrival_Date'].max() - wd['Arrival_Date'].min())
```

```python
#Descriptive Statistics
wd[['Cargo_tons', 'Berth_Time_hr']].agg(['mean', 'median','std', 'skew', 'kurt'])
```

```python
# investigate if there are duplicates
print(wd.duplicated().sum()) # no duplicated entries 
print(wd.duplicated(subset=["Vessel_ID"]).sum())


# investigate if there are missing values
print(wd.isnull().sum().sum()) # no missing data 
print(wd.isnull().sum()[wd.isnull().sum() > 0])
print(wd.dtypes)
print(wd.describe)

# Consitency check 
print((wd['Berth_Time_hr'] < 0).sum())
print((wd['Cargo_tons'] < 0).sum())

# check if each vessel ID has one unit corresponding vessel type
wd.groupby('Vessel_ID')['Vessel_Type'].nunique().value_counts()
```

```python
# 4.1 investigate more with plotting the data 

fig, ax = plt.subplots(1, 2, figsize=(11, 4))

wd['Cargo_tons'].hist(bins=40, ax=ax[0])
ax[0].set_title('Cargo_tons distribution')
ax[0].set_xlabel('Cargo (tons)')
ax[0].set_ylabel('Count')

wd['Berth_Time_hr'].hist(bins=40, ax = ax[1])
ax[1].set_title('Berth_time_hr')
ax[1].set_xlabel('Berth time (hours)')
ax[1].set_ylabel('Count')
```
